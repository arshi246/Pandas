# Pandas DataFrame Core Operations

Pandas is a Python library used for **working with structured and tabular data**.

It is very useful in:

- Data Science
- Machine Learning
- Data Analysis
- Artificial Intelligence
- Data Cleaning

The main data structure of Pandas is called a **DataFrame**.

A DataFrame stores data in the form of:
```
Rows + Columns
```

For example:

```
      Name    Age    Marks
0     Ali     20      80
1     Sara    21      90
2     Ahmed   19      75
```

We can imagine a DataFrame like an **Excel table**.

# 14. Handling Missing Values

Real datasets often contain missing data.

NaN means a value is missing.

Example:


````python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print(df)
````

Check Missing Values



````python
print(df.isnull())
````

To count missing values:

````python
print(df.isnull().sum())
````

Remove Missing Values

dropna() removes rows containing missing values.

````python
df = df.dropna()
df
````

## Fill Missing Values

Instead of deleting them:

````python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

df['Marks'] = df['Marks'].fillna(0)
df
````

We can also fill missing marks using average marks:

````python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

df['Marks'] = df['Marks'].fillna(df['Marks'] .mean())
df
````

## groupby()


groupby() is very important in Data Analysis.



`groupby()` allows us to analyze data according to categories.

````python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Course": ["AI", "Web", "AI", "Web"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)
df
````

# 18. Combining DataFrames

Sometimes our data exists in multiple DataFrames.

Pandas allows us to combine them.

## concat()


concat() is useful when we want to join similar datasets vertically.




Suppose:

````python
df1 =pd.DataFrame({
    "Name" : ["Ali","Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
    })

````

Combine them:

````python

final_df = pd.concat([df1,df2], ignore_index=True)

print(final_df)
````

## merge()

`merge()` combines DataFrames using a common column.

Example:

````python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})
````

Merge using ID:

````python
final_df = pd.merge(students,marks ,on = 'ID')
print(final_df)
````

The common column is:

```
ID
```

So:

```
Students Data
      +
Marks Data
      ↓
Common ID
      ↓
Final DataFrame
```

# Class Practice: Student Performance Analyzer

Now we will use different Pandas concepts in one small program.

## Problem

We have information about different students.

We want to find:

- Average marks
- Highest marks
- Lowest marks
- Passed students
- Failed students
- Course-wise average
- Add a Result column

````python
data = {
    "Name" : ["Ali","Sara","Ahmed","Ayesha","Zain"],
    "Course" : ["AI", "Web", "AI","Web","AI"],
    "Marks" : [45,90, 75,85, 40]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print(f"\nAverage Marks : {df['Marks'].mean()}")
print(f"Highest Marks : {df['Marks'].max()}")
print(f"Lowest Marks : {df['Marks'].min()}")


passed = df[df['Marks'] >=50]
failed = df[df['Marks'] <50]


print(f"\nPassed students : {passed}")
print(f"Failed students : {failed}")

df["Result"] = df['Marks'].apply(lambda marks : "Pass" if marks >= 50 else "Fail")



print("\nFinal Data:")
print(df)


course_average = df.groupby("Course")["Marks"].mean()
print(f"\nCourse Wise Average : {course_average}")



````
